## Linear filtering by convolution (mean filter)

**<a id="#">Task 3.1.1.</a>** Load the image `data/astronaut.png` and show it:

**<a id="#">Task 3.1.2.</a>** Implement a mean filter:

```python
def meanfilter(img_in, size):
    result = numpy.zeros(img_in.shape)
    n = (size - 1) // 2
    for p in numpy.ndindex(img_in.shape):
        neighborhood_sum = 0
        count = 0
        if p[0] >=n and p[1] >=n and p[0]< img_in.shape[0] - n and p[1] < img_in.shape[0] - n:   
            for q in numpy.ndindex(size, size):
                i = p[0] + q[0] - n
                j = p[1] + q[1] - n
                neighborhood_sum += img_in[i, j]
                count += 1
            result[p] = neighborhood_sum / count
    return result

img_2= img.copy()
image_mean=meanfilter(img_2, size = 5)
plt.imshow(image_mean)
```

## Non-linear filtering

**<a id="#">Task 3.2.1.</a>** Implement a median filter:

```python
def medianfilter(img_in, size):
    result = numpy.zeros(img_in.shape)
    n = (size - 1) // 2
    for p in numpy.ndindex(img_in.shape):
        neighborhood = []
        if p[0] >=n and p[1] >=n and p[0]< img_in.shape[0] - n and p[1] < img_in.shape[0] - n:   
            for q in numpy.ndindex(size, size):
                i = p[0] + q[0] - n
                j = p[1] + q[1] - n
                neighborhood.append(img_in[i, j])
            result[p] = numpy.median(neighborhood)
    return result
```
